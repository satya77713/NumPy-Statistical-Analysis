```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/sales_data.csv")
```

```python
display(df.head())
```

```python
data = pd.to_numeric(df["Sales_Amount"], errors="coerce")
data = data.dropna().to_numpy()
```

```python
mean_value = np.mean(data)
print(mean_value)
```

```python
median_value = np.median(data)
print(median_value)
```

```python
std_value = np.std(data)
print(std_value)
```

```python
min_value = np.min(data)
print(min_value)
```

```python
max_value = np.max(data)
print(max_value)
```

```python
if mean_value > median_value:
    print("Mean is greater than median.")
elif mean_value < median_value:
    print("Mean is less than median.")
else:
    print("Mean and median are equal.")
```

```python
low_sales = data[data < 1000]
high_sales = data[data >= 4500]

print("\nSales below 1000:", len(low_sales))
print("Sales of 4500 or more:", len(high_sales))

print("\nSample low sales values:", low_sales[:10])
print("Sample high sales values:", high_sales[:10])
```
